## Atividade 4 - Escalonamento de variáveis numéricas (StandardScaler e MinMaxScaler)

### Dataset 1
Link: archive.ics.uci.edu/ml/datasets/adult 

```python
adult = fetch_ucirepo(id=2) 
  
X = adult.data.features 
y = adult.data.targets 
  
df_adult = pd.concat([X, y], axis=1)
```

```python
df_adult.head()
```

```python
num_cols = ['age', 'fnlwgt', 'capital-gain', 'capital-loss', 'hours-per-week']
```

```python
df_num = df_adult[num_cols].copy()
```

```python
scaler_std = StandardScaler()
df_std = pd.DataFrame(
    scaler_std.fit_transform(df_num),
    columns=[f'padro_{col}' for col in num_cols],
    index=df_adult.index
)
```

```python
scaler_minmax = MinMaxScaler()
df_minmax = pd.DataFrame(
    scaler_minmax.fit_transform(df_num),
    columns=[f'norm_{col}' for col in num_cols],
    index=df_adult.index
)
```

```python
df_final = pd.concat([df_std, df_minmax], axis=1)
```

```python
df_final
```
