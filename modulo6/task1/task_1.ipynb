```python
from ucimlrepo import fetch_ucirepo 
import pandas as pd
```

## Dataset 1
Link: archive.ics.uci.edu/ml/datasets/adult 

```python
adult = fetch_ucirepo(id=2) 
  
X = adult.data.features 
y = adult.data.targets 
  
df_adult = pd.concat([X, y], axis=1)
```

```python
df_adult.head()
```

```python
def identify_invalid_rows(df):
    invalid_conditions = (
        (df["age"] <= 0) | (df["age"] > 120) |
        (df["fnlwgt"] <= 0) |
        (df["education-num"] < 1) | (df["education-num"] > 16) |
        (df["capital-gain"] < 0) |
        (df["capital-loss"] < 0) |
        (df["hours-per-week"] <= 0) | (df["hours-per-week"] > 100)
    )
    return df[invalid_conditions]
```

```python
invalid_rows = identify_invalid_rows(df_adult)
df_adult_clean = df_adult.drop(index=invalid_rows.index).reset_index(drop=True)
```

```python
df_adult_clean
```
