```python
from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder
from sklearn.compose import ColumnTransformer
from ucimlrepo import fetch_ucirepo 
import pandas as pd
```

## Atividade 3 - Codificação de variáveis categóricas (OrdinalEncoder e OneHotEncoder)

### Dataset 1
Link: archive.ics.uci.edu/ml/datasets/adult 

```python
adult = fetch_ucirepo(id=2) 
  
X = adult.data.features 
y = adult.data.targets 
  
df_adult = pd.concat([X, y], axis=1)
```

```python
df_adult.head()
```

```python
ordinal_cols = ['education']
nominal_cols = ['workclass', 'marital-status', 'occupation',
                'relationship', 'race', 'sex', 'native-country', 'income']
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ('ord', OrdinalEncoder(), ordinal_cols),
        ('nom', OneHotEncoder(handle_unknown='ignore', sparse_output=False), nominal_cols)
    ],
    remainder='passthrough'
)
```

```python
df_transformed = preprocessor.fit_transform(df_adult)
```

```python
all_columns = preprocessor.get_feature_names_out()
```

```python
df_adult_encoded = pd.DataFrame(df_transformed, columns=all_columns, index=df_adult.index)
```

```python
df_adult_encoded
```

### Dataset 3
Link: archive.ics.uci.edu/ml/datasets/credit+approval

```python
credit_approval = fetch_ucirepo(id=27) 
  
X = credit_approval.data.features 
y = credit_approval.data.targets 
  
df_credit_approval = pd.concat([X, y], axis=1)
```

```python
df_credit_approval
```

```python
ordinal_cols = ['A13', 'A12', 'A10', 'A9', 'A8', 'A7', 'A6', 'A4', 'A3', 'A2', 'A1']
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ('ord', OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1), ordinal_cols)
    ],
    remainder='drop'
)
```

```python
df_transformed = preprocessor.fit_transform(df_credit_approval)
```

```python
ordinal_names = ordinal_cols
all_columns = list(ordinal_names)
```

```python
df_credit_approval_encoded = pd.DataFrame(df_transformed, columns=all_columns, index=df_credit_approval.index)
```

### Dataset 4
Link: https://www.kaggle.com/code/kerneler/starter-ames-housing-fb60970c-d

```python
df_ames_housing = pd.read_excel('./datasets/AmesHousing.xls', engine="xlrd")
```

```python
df_ames_housing.isnull().sum()
```

```python
df_ames_housing.columns
```

```python
ordinal_cols = [
    'Overall Qual', 'Overall Cond', 'Exter Qual', 'Exter Cond', 'Bsmt Qual',
    'Bsmt Cond', 'Heating QC', 'Kitchen Qual', 'Fireplace Qu', 'Garage Qual',
    'Garage Cond', 'Pool QC', 'Fence', 'Misc Feature'
]

nominal_cols = [
    'MS Zoning', 'Street', 'Alley', 'Lot Shape', 'Land Contour', 'Utilities',
    'Lot Config', 'Land Slope', 'Neighborhood', 'Condition 1', 'Condition 2',
    'Bldg Type', 'House Style', 'Roof Style', 'Roof Matl', 'Exterior 1st',
    'Exterior 2nd', 'Mas Vnr Type', 'Foundation', 'Bsmt Exposure',
    'BsmtFin Type 1', 'BsmtFin Type 2', 'Heating', 'Central Air', 'Electrical',
    'Functional', 'Garage Type', 'Garage Finish', 'Paved Drive', 'Sale Type',
    'Sale Condition'
]
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ('ord', OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1), ordinal_cols),
        ('nom', OneHotEncoder(handle_unknown='ignore', sparse_output=False), nominal_cols)
    ],
    remainder='passthrough'  
)
```

```python
df_encoded = preprocessor.fit_transform(df_ames_housing)
```

```python
ordinal_names = ordinal_cols
nominal_names = preprocessor.named_transformers_['nom'].get_feature_names_out(nominal_cols)
other_cols = [c for c in df_ames_housing.columns if c not in ordinal_cols + nominal_cols]
all_columns = list(ordinal_names) + list(nominal_names) + other_cols
```

```python
df_ames_housing_encoded = pd.DataFrame(df_encoded, columns=all_columns, index=df_ames_housing.index)
```
