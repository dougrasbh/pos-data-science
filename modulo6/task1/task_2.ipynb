## Atividade 2 - Imputação de valores faltantes

### Dataset 1
Link: archive.ics.uci.edu/ml/datasets/adult 

```python
adult = fetch_ucirepo(id=2) 
  
X = adult.data.features 
y = adult.data.targets 
  
df_adult = pd.concat([X, y], axis=1)
```

```python
df_adult.head()
```

```python
df_adult.isnull().sum()
```

```python
cat_cols = df_adult.select_dtypes(exclude='number').columns
df_adult[cat_cols] = df_adult[cat_cols].fillna("None")
```

```python
df_adult.isnull().sum()
```

### Dataset 3
Link: archive.ics.uci.edu/ml/datasets/credit+approval

```python
credit_approval = fetch_ucirepo(id=27) 
  
X = credit_approval.data.features 
y = credit_approval.data.targets 
  
df_credit_approval = pd.concat([X, y], axis=1)
```

```python
df_credit_approval
```

```python
df_credit_approval.isnull().sum()
```

#### Inserindo dados faltantes com a média

```python
df_credit_approval_mean = df_credit_approval.copy()
```

```python
num_cols = df_credit_approval_mean.select_dtypes(include='number').columns

for col in num_cols:
  mean_value = df_credit_approval_mean[col].mean()
  print(f"Coluna: {col}, Média: {mean_value}")
  df_credit_approval_mean[col].fillna(mean_value, inplace=True)
```

```python
df_credit_approval_mean[df_credit_approval_mean['A14'] == 184.01477104874445]
```

#### Inserindo dados faltantes com a moda

```python
df_credit_approval_mode = df_credit_approval.copy()
```

```python
num_cols = df_credit_approval_mode.select_dtypes(include='number').columns

for col in num_cols:
  mode_value = df_credit_approval_mode[col].mode()[0]
  print(f"Coluna: {col}, Moda: {mode_value}")
  df_credit_approval_mode[col].fillna(mode_value, inplace=True)
```

```python
df_credit_approval_mean[df_credit_approval_mean['A14'] == 0]
```
